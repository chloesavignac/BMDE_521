```python
def preprocess_file(adata):
    """
    Perform basic quality control (QC) filtering on single-cell RNA-seq data.

    Filtering criteria:
        - Retain cells with at least 200 detected genes.
        - Retain genes detected in at least 3 cells.
        - Exclude cells with 2,500 or more detected genes (potential doublets).
        - Exclude cells with 5% or more mitochondrial counts (indicator of cellular damage or stress).

    Parameters
    ----------
    adata : anndata.AnnData
        Annotated single-cell gene expression matrix containing raw counts.
        Mitochondrial genes are expected to start with 'MT-'.

    Returns
    -------
    anndata.AnnData
        Filtered AnnData object with updated QC metrics.
    """

    print(f'{adata.n_obs} cells before QC')
    print('Verifying the object stored the raw counts.')
    print(f'Expression range: ({adata.X.min()}, {adata.X.max()})')

    # Filter cells and genes based on minimum detection thresholds
    sc.pp.filter_cells(adata, min_genes=200)
    sc.pp.filter_genes(adata, min_cells=3)

    print(f'Number of genes kept: {len(adata.var)}')

    # Exclude cells with unusually high numbers of detected genes
    # (potential doublets)
    adata = adata[adata.obs.n_genes < 2500, :].copy()

    print(f'{adata.n_obs} cells passed the first round of QC')

    # Identify mitochondrial genes
    adata.var['mt'] = adata.var_names.str.startswith('MT-')

    # Calculate QC metrics, including the percentage of mitochondrial counts
    sc.pp.calculate_qc_metrics(
        adata,
        qc_vars=['mt'],
        percent_top=None,
        log1p=False,
        inplace=True
    )

    # Exclude cells with high mitochondrial content
    adata = adata[adata.obs.pct_counts_mt < 5, :].copy()

    print(f'{adata.n_obs} cells remaining after mitochondrial filtering')

    # Inspection
    sc.pl.scatter(
        adata,
        x='total_counts',
        y='n_genes_by_counts',
        color='pct_counts_mt'
    )

    # Divides the expression of each gene by the total counts detected in that cell
    sc.pp.normalize_total(adata, target_sum=1e4) # multiplies the result by 10,000
    sc.pp.log1p(adata) # log-normalization (most genes have low expression, the distribution is skewed)

    return adata
```

### NU02954_PBMC

```python
preprocess_file(NU02954_PBMC) # Running once without saving

# There is an anomaly here... cells with high total counts, but low_genes_by counts (dark population in first plot below)

NU02954_PBMC.var['mt'] = NU02954_PBMC.var_names.str.startswith('MT-')

# Additional inspection
sc.pp.calculate_qc_metrics(
    NU02954_PBMC,
    qc_vars=['mt'],
    percent_top=[10, 20, 50],
    log1p=False,
    inplace=True
)

sc.pl.scatter(
    NU02954_PBMC,
    x='total_counts',
    y='n_genes_by_counts',
    color='pct_counts_in_top_20_genes'
)

# Identify cells with high counts but low gene complexity
suspicious_cells = NU02954_PBMC[
    (NU02954_PBMC.obs.total_counts > 3000) &
    (NU02954_PBMC.obs.n_genes_by_counts < 600)
].copy()

# Identify their most highly expressed genes
sc.pl.highest_expr_genes(
    suspicious_cells,
    n_top=20
)
```

```python
# HBB accounts for approximately 80–85% of total counts in many of the cells in the NU02954_PBMC sample 
# These are erythrocyte-associated markers (red blood cells)
# Mature erythrocytes do not have a nucleus, so this is probably a sign of contamination

adata = NU02954_PBMC.copy()

# Calculate HBB percentage per cell
hbb = adata[:, 'HBB'].X

if hasattr(hbb, 'toarray'):
    hbb = hbb.toarray()

hbb = np.asarray(hbb).flatten()

total_counts = np.asarray(adata.X.sum(axis=1)).flatten()

adata.obs['pct_counts_HBB'] = np.divide(
    hbb,
    total_counts,
    out=np.zeros_like(hbb, dtype=float),
    where=total_counts > 0
) * 100

# Visualize
sc.pl.scatter(
    adata,
    x='total_counts',
    y='n_genes_by_counts',
    color='pct_counts_HBB',
    color_map='viridis'
)

sc.pl.violin(
    adata,
    keys=['pct_counts_HBB'],
    jitter=0.4
)

# Additional filtering for that object based on the extra QC steps

# We can see in the plot below that they are 2 distributions 
# We will select cells with low HBB counts 

NU02954_PBMC = adata[adata.obs.pct_counts_HBB < 30, :].copy()
```

```python
# Verifying that the stored counts are log-normalized 
NU02954_PBMC_proccessed.X.max(), NU02954_Tumor_processed.X.max(), NU03014_PBMC_processed.X.max(), NU03014_Tumor_processed.X.max()
```

```python
# Before integration, let's look at the number of overlapping genes

datasets = {
    'NU02954_PBMC': NU02954_PBMC_proccessed,
    'NU02954_Tumor': NU02954_Tumor_processed,
    'NU03014_PBMC': NU03014_PBMC_processed,
    'NU03014_Tumor': NU03014_Tumor_processed
}

for name, adata in datasets.items():
    print(f'{name}: {adata.n_obs} cells × {adata.n_vars} genes')

gene_sets = [set(adata.var_names) for adata in datasets.values()]

shared_genes = set.intersection(*gene_sets)
all_genes = set.union(*gene_sets)

print(f'Genes shared across all datasets: {len(shared_genes)}')
print(f'Total unique genes: {len(all_genes)}')
```

```python
# Integrating datasets
adata_combined = ad.concat(
    [NU02954_PBMC_proccessed, NU02954_Tumor_processed, NU03014_PBMC_processed, NU03014_Tumor_processed],
    axis=0,
    join='inner',
    label='batch',
    keys=['NU02954_PBMC', 'NU02954_Tumor', 'NU03014_PBMC', 'NU03014_Tumor'],
    index_unique='-'
)

# Extract patient and tissue information
adata_combined.obs['patient'] = (
    adata_combined.obs['batch'].str.split('_').str[0]
)

adata_combined.obs['tissue'] = (
    adata_combined.obs['batch'].str.split('_').str[1]
)
```

```python
sc.pp.highly_variable_genes(adata_combined, min_mean=0.0125, max_mean=3, min_disp=0.5)
sc.pl.highly_variable_genes(adata_combined)

# A negative value indicates that a gene is less variable than expected relative to genes with comparable mean expression.
```

## Creating a new label combined both scanpy and mapmycells

```python
# Initialize with a copy of obs
adata_combined_processed_obs = adata_combined_processed.obs.copy()

# Updating obs with cell annotations (left join to preserve all cells)
adata_combined_processed_obs = adata_combined_processed_obs.merge(
    cell_annotations,
    left_index=True,
    right_index=True,
    how='left',
    validate='one_to_one'   # raises if the index isn't unique on either side
)

# Sanity check: how many cells failed to get an annotation?
n_missing = adata_combined_processed_obs['cluster_name_Siletti'].isna().sum()
print(f"{n_missing} cells have no matching annotation after merge")

# Initialize as object dtype so string assignment doesn't trigger dtype warnings
adata_combined_processed_obs['combined_annotation'] = pd.Series(
    [None] * len(adata_combined_processed_obs),
    index=adata_combined_processed_obs.index,
    dtype='object'
)

# Comparing with Scanpy labels
for cl in adata_combined_processed_obs["cell_type_lvl1"].dropna().unique():

    print(cl)

    cl_cells = adata_combined_processed_obs.loc[
        adata_combined_processed_obs["cell_type_lvl1"] == cl
    ].copy()

    # Filter by bootstrapping probability
    cl_cells_annot_confident = cl_cells.loc[
        cl_cells.cluster_bootstrapping_probability > 0.50
    ].copy()

    # Calculate proportion of each predicted cluster
    mapping_results = (
        cl_cells_annot_confident.cluster_name_Siletti
        .value_counts(normalize=True)
        .to_frame(name='proportion')
    )

    # Retain predicted clusters above 20%
    mapping_results = mapping_results.loc[
        mapping_results['proportion'] > 0.2
    ]

    print(mapping_results)
    print('\n')

    # Identify cells belonging to retained clusters
    mask = cl_cells_annot_confident.cluster_name_Siletti.isin(
        mapping_results.index
    )

    selected_cells = cl_cells_annot_confident.loc[mask]

    # Assign combined annotations
    adata_combined_processed_obs.loc[
        selected_cells.index, 'combined_annotation'
    ] = (
        selected_cells['cell_type_lvl1'].astype(str)
        + '_' +
        selected_cells['cluster_name_Siletti'].astype(str)
    )

    # Fill any remaining NaNs with the level 1 annotation
    adata_combined_processed_obs['combined_annotation'] = (
        adata_combined_processed_obs['combined_annotation']
        .fillna(adata_combined_processed_obs['cell_type_lvl1'])
    )
```

```python
from gseapy import barplot, dotplot

gene_list = list(adata_combined_processed.var.index) # we will consider all genes in our pathway analysis
print(gene_list[0:10]) # verify format of gene names

# Loop through each cluster/cell type combination and perform enrichment analysis
for cell_cluster in list(counts_df.index):

    print(cell_cluster)

    cluster_subset = adata_combined_processed[adata_combined_processed.obs.combined_annotation==cell_cluster, :]

    enr = gp.enrichr(
        gene_list=gene_list,                              
        gene_sets=['MSigDB_Hallmark_2020', 'KEGG_2021_Human'], # Hallmark is broad, KEGG is more specific
        organism='human',                               
        outdir=None,
    )

    # Keep only statistically significant results
    sig_enr = enr.res2d[enr.res2d['Adjusted P-value']<0.05].reset_index(drop=True)

    print(sig_enr.Gene_set.value_counts()) # see number of enrichment in each atlas

    ax = dotplot(
        sig_enr,
        column="Adjusted P-value",
        x='Gene_set',          # split the x-axis by library (multi-library comparison)
        size=3,
        top_term=10,
        figsize=(3, 8),
        title=f"{cell_cluster}",
        xticklabels_rot=45,    # rotate x tick labels for readability
        show_ring=True,        # set False to remove the outer ring around dots
        marker='o',
    )

    ax = barplot(
        enr.results,
        column="Adjusted P-value",
        group='Gene_set',      # group bars by library (multi-library comparison)
        size=5,
        top_term=5,
        figsize=(3, 5),
        title=f"{cell_cluster}",
        color={'KEGG_2021_Human': 'salmon', 'MSigDB_Hallmark_2020': 'darkblue'},
    )

    plt.show()
    plt.close('all')

```
